# Exponential Decay

## What Is It?
Exponential decay describes any system where the rate of
decrease is proportional to the current value.

**Real world examples:**
- Radioactive decay
- Drug leaving your bloodstream
- Coffee cooling down

## The ODE
`dy/dt = -y,  y(0) = 1`

**Exact solution:** `y(t) = e^(-t)`

```python
import sys, os
sys.path.insert(0, os.path.join('..', 'python'))
sys.path.insert(0, os.path.join('..', 'build'))
import numpy as np
import matplotlib.pyplot as plt
from fastode_interface import solve
print('FastODE loaded!')
```

```python
def f(t, y):
    return [-y[0]]

result = solve(f, t0=0.0, t1=5.0, y0=[1.0], method='RK45')
t_exact = np.linspace(0, 5, 200)
y_exact = np.exp(-t_exact)

print(f'Steps taken:    {result.n_steps}')
print(f'Final value:    {result.y[0,-1]:.8f}')
print(f'Exact value:    {np.exp(-5):.8f}')
print(f'Error:          {abs(result.y[0,-1] - np.exp(-5)):.2e}')
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(t_exact, y_exact, 'b-', linewidth=2, label='Exact')
axes[0].plot(result.t, result.y[0], 'ro--', markersize=8,
             label=f'FastODE ({result.n_steps} steps)')
axes[0].set_xlabel('Time')
axes[0].set_ylabel('y(t)')
axes[0].set_title('Exponential Decay')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

y_interp = np.interp(result.t, t_exact, y_exact)
error = np.abs(result.y[0] - y_interp)
axes[1].semilogy(result.t, error + 1e-20, 'g-o', markersize=6)
axes[1].set_xlabel('Time')
axes[1].set_ylabel('Absolute Error')
axes[1].set_title('Error vs Exact Solution')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('exponential_decay.png', dpi=100)
plt.show()
print('Plot saved!')
```

```python
rk4  = solve(f, t0=0, t1=5, y0=[1.0], method='RK4',  h=0.5)
rk45 = solve(f, t0=0, t1=5, y0=[1.0], method='RK45')

plt.figure(figsize=(8, 4))
plt.plot(t_exact, y_exact, 'b-', linewidth=2, label='Exact')
plt.plot(rk4.t,  rk4.y[0],  'ro-', markersize=10,
         label=f'RK4  ({rk4.n_steps} steps, h=0.5)')
plt.plot(rk45.t, rk45.y[0], 'gs-', markersize=8,
         label=f'RK45 ({rk45.n_steps} steps, adaptive)')
plt.xlabel('Time')
plt.ylabel('y(t)')
plt.title('RK4 vs RK45 Comparison')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
print(f'RK4  used {rk4.n_steps} steps')
print(f'RK45 used {rk45.n_steps} steps')
print(f'RK45 is {rk4.n_steps/rk45.n_steps:.1f}x more efficient')
```
